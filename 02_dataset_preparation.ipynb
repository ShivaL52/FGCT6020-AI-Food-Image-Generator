```python
from torchvision.datasets import Food101

# Downloads and unpacks the full Food-101 dataset into Colab (not into Drive)
Food101(root="/content/data", split="train", download=True)
print("Download finished")
```

```python
import os, shutil

src = "/content/data/food-101"
DRIVE = "/content/drive/MyDrive/FGCT6020_Project for AI for Creativity/02_dataset"

for split in ["train", "test"]:
    # Read the official list of images for this split
    with open(f"{src}/meta/{split}.txt") as f:
        names = [line.strip() for line in f if line.startswith("chicken_curry/")]

    out = f"{DRIVE}/raw_{split}"
    os.makedirs(out, exist_ok=True)

    for n in names:
        shutil.copy(f"{src}/images/{n}.jpg", f"{out}/{os.path.basename(n)}.jpg")

    print(split, "images copied:", len(names))
```
